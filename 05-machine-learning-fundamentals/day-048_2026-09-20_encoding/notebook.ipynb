## Objective
Encoding converts categories into numeric features. One-hot encoding avoids imposing a false order on nominal categories.

## Practice task
One-hot encode a nominal category and confirm unseen categories do not break transformation.


```python
import pandas as pd
from sklearn.preprocessing import OneHotEncoder

categories = pd.DataFrame({"city": ["Agra", "Delhi", "Noida", "Agra"]})
encoder = OneHotEncoder(handle_unknown="ignore", sparse_output=False)
encoded = encoder.fit_transform(categories[["city"]])
print("Categories:", encoder.categories_)
print(pd.DataFrame(encoded, columns=encoder.get_feature_names_out(["city"])))
```
