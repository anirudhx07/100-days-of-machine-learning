## Objective
A perceptron computes a weighted sum and applies a threshold. It can learn linearly separable classes but cannot solve XOR alone.

## Practice task
Implement a perceptron update rule and test it on linearly separable AND data.


```python
import numpy as np

X = np.array([[0, 0], [0, 1], [1, 0], [1, 1]], dtype=float)
y = np.array([0, 0, 0, 1])  # AND is linearly separable
weights = np.zeros(2); bias = 0.0; learning_rate = 0.1
for epoch in range(10):
    errors = 0
    for features, target in zip(X, y):
        prediction = int(features @ weights + bias >= 0.5)
        update = learning_rate * (target - prediction)
        weights += update * features; bias += update
        errors += int(update != 0)
    print(f"epoch={epoch + 1}, errors={errors}")
print("Predictions:", (X @ weights + bias >= 0.5).astype(int))
```
