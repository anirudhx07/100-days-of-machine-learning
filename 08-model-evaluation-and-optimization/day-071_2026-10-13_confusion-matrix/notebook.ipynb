## Objective
A confusion matrix counts true and predicted class pairs. Its rows and columns must be labeled to avoid interpreting errors backward.

## Practice task
Create a labeled confusion matrix and identify false positives and false negatives.


```python
from sklearn.metrics import confusion_matrix, ConfusionMatrixDisplay
import matplotlib.pyplot as plt

y_true = [1, 0, 1, 1, 0, 0, 1, 0]
y_pred = [1, 0, 0, 1, 0, 1, 1, 0]
cm = confusion_matrix(y_true, y_pred, labels=[0, 1])
print("Rows=true [0,1], columns=predicted [0,1]:\n", cm)
ConfusionMatrixDisplay(cm, display_labels=["Negative", "Positive"]).plot()
plt.tight_layout()
plt.show()
```
