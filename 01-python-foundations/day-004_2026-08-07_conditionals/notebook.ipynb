## Practice task
Write a grading function with explicit boundaries for A, B, C, and fail; test every boundary value.


```python
def grade(score: float) -> str:
    if not 0 <= score <= 100:
        raise ValueError("score must be between 0 and 100")
    if score >= 90:
        return "A"
    elif score >= 75:
        return "B"
    elif score >= 60:
        return "C"
    return "Fail"

for score in [95, 82, 64, 42, 75, 60]:
    print(score, "->", grade(score))
```
