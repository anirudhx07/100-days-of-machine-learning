## Practice task
Build a bill calculator using arithmetic operators and comparisons; test zero and negative inputs.


```python
price = 799.0
discount_percent = 15
tax_percent = 18

discounted = price * (1 - discount_percent / 100)
final_price = discounted * (1 + tax_percent / 100)
print(f"Original: ₹{price:.2f}")
print(f"Final after discount and tax: ₹{final_price:.2f}")

quantity = 2
print("Bulk order:", quantity >= 2 and final_price > 500)
```
