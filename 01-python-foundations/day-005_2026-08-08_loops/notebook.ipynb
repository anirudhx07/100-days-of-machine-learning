## Practice task
Use a loop to calculate the sum and average of a list without calling sum() or statistics.mean().


```python
numbers = [12, 7, 19, 4, 8]
total = 0
for value in numbers:
    total += value
print("Total:", total)
print("Average:", total / len(numbers))

countdown = 3
while countdown > 0:
    print(countdown)
    countdown -= 1
print("Go!")
```
