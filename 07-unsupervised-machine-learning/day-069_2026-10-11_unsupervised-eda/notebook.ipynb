## Practice task
Create a two-dimensional projection and compare cluster sizes and feature distributions.


```python
import pandas as pd
import seaborn as sns
import matplotlib.pyplot as plt
from sklearn.datasets import load_iris
from sklearn.preprocessing import StandardScaler
from sklearn.cluster import KMeans

data = load_iris(as_frame=True)
df = data.frame.copy()
features = list(data.feature_names)
X = StandardScaler().fit_transform(df[features])
df["cluster"] = KMeans(n_clusters=3, n_init=10, random_state=42).fit_predict(X)
print(df.groupby("cluster")[features].mean().round(2))
sns.pairplot(df, vars=features[:3], hue="cluster", corner=True)
plt.show()
```
