## Objective
Silhouette score compares within-cluster cohesion with nearest-cluster separation. It is useful but does not prove that clusters are meaningful.

## Practice task
Calculate silhouette score for a clustering result and describe one limitation of the score.


```python
from sklearn.datasets import make_blobs
from sklearn.preprocessing import StandardScaler
from sklearn.cluster import KMeans
from sklearn.metrics import silhouette_score

X, _ = make_blobs(n_samples=300, centers=4, cluster_std=0.9, random_state=5)
X_scaled = StandardScaler().fit_transform(X)
labels = KMeans(n_clusters=4, n_init=10, random_state=5).fit_predict(X_scaled)
print("Silhouette score:", round(silhouette_score(X_scaled, labels), 3))
```
