## Practice task
Create a 2D array, reshape it without changing element count, and assert the new shape.


```python
import numpy as np

arr = np.arange(12)
matrix = arr.reshape(3, 4)
print("Original:", arr)
print("Reshaped:\n", matrix)
print("Transpose shape:", matrix.T.shape)
print("Zeros:\n", np.zeros((2, 3)))
```
