## Practice task
Create a NumPy array from a list and report its dtype, shape, size, and number of dimensions.


```python
import numpy as np

values = [2, 4, 6, 8]
arr = np.array(values)
print("Array:", arr)
print("dtype:", arr.dtype, "shape:", arr.shape, "size:", arr.size, "ndim:", arr.ndim)
print("Doubled:", arr * 2)
```
