Probability quantifies uncertainty from 0 to 1. For equally likely outcomes, probability is favorable outcomes divided by total outcomes.

## Practice task
Simulate 10,000 coin flips and compare empirical head probability with 0.5.


```python
import numpy as np

rng = np.random.default_rng(42)
flips = rng.choice([0, 1], size=10_000)  # 1 represents heads
heads = int(flips.sum())
print("Heads:", heads)
print("Empirical P(heads):", heads / len(flips))
print("Theoretical P(heads):", 0.5)
```
