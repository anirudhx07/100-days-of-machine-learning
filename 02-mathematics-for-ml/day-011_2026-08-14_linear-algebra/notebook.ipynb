## Practice task
Represent two small matrices and verify the result shape of their multiplication.


```python
import numpy as np

A = np.array([[1, 2], [3, 4]], dtype=float)
B = np.array([[5, 6], [7, 8]], dtype=float)
print("A shape:", A.shape)
print("A + B:\n", A + B)
print("A @ B:\n", A @ B)
print("Transpose:\n", A.T)
```
