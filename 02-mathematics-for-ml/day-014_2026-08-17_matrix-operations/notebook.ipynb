## Objective
Matrix addition is element-wise; matrix multiplication combines rows and columns. The inner dimensions must match: (m, n) @ (n, p) -> (m, p).

## Practice task
Compare matrix multiplication with element-wise multiplication and explain the difference in comments.


```python
import numpy as np

A = np.array([[1, 2, 3], [4, 5, 6]])  # (2, 3)
B = np.array([[1, 2], [3, 4], [5, 6]])  # (3, 2)
print("Matrix product shape:", (A @ B).shape)
print(A @ B)
print("Element-wise product is only valid for matching/broadcastable shapes:")
C = np.array([[10, 10, 10], [20, 20, 20]])
print(A * C)
```
