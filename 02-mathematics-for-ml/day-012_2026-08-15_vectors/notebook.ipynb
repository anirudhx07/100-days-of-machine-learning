## Practice task
Compute the dot product of two vectors manually and with NumPy; verify the results match.


```python
import numpy as np

a = np.array([2, 3, 4])
b = np.array([1, 0, -1])
manual_dot = sum(x * y for x, y in zip(a, b))
print("Manual dot product:", manual_dot)
print("NumPy dot product:", np.dot(a, b))
print("Magnitude of a:", np.linalg.norm(a))
```
