## Practice task
Create a 3x3 matrix and extract its center element, first row, and last column.


```python
import numpy as np

M = np.arange(1, 10).reshape(3, 3)
print("Matrix:\n", M)
print("Center:", M[1, 1])
print("First row:", M[0, :])
print("Last column:", M[:, -1])
print("Shape:", M.shape, "Dimensions:", M.ndim)
```
