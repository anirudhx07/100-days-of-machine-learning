## Objective
Variance measures average squared distance from the mean; standard deviation is its square root and returns to the original unit.

## Practice task
Calculate sample and population standard deviation and explain why they differ.


```python
import numpy as np

data = np.array([4, 5, 6, 7, 8], dtype=float)
print("Mean:", data.mean())
print("Population variance:", data.var(ddof=0))
print("Sample variance:", data.var(ddof=1))
print("Population std:", data.std(ddof=0))
print("Sample std:", data.std(ddof=1))
```
